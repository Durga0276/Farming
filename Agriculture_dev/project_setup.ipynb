```python
catalog = "agriculture_dev"
schemas = ["Source", "bronze", "silver", "gold"]
volume_schema = "Source"
volume = "agriculture_dataset"

# 1) Catalog
spark.sql(f"CREATE CATALOG IF NOT EXISTS `{catalog}`")

# 2) Schemas (Source for landing files, bronze/silver/gold for medallion layers)
for schema in schemas:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`")

# 3) Managed volume for the source CSV
spark.sql(f"CREATE VOLUME IF NOT EXISTS `{catalog}`.`{volume_schema}`.`{volume}`")

#upload agriculture dataset in created volume
```
